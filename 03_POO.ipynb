Devant une question, on ne commence **jamais** par écrire du code. On suit ces 5 étapes, sur le brouillon :

| Étape | Ce que je fais | Ce que j'écris au brouillon |
|:---|:---|:---|
| **1. Lire** | Je lis la question **et** les lignes juste au-dessus (souvent, elles donnent la solution en français). Je souligne les verbes : *écrire, compléter, donner, expliquer, justifier*. | les mots soulignés |
| **2. Reconnaître le type** | Je cherche dans le tableau ci-dessus à quel type ressemble la question. | « c'est un type … » |
| **3. Appliquer la fiche** | J'ouvre la fiche du type (plus bas) et je suis sa méthode pas à pas. | le tableau de trace, l'arbre, le squelette de code |
| **4. Vérifier à la main** | Je teste ma réponse sur **l'exemple de l'énoncé** (il y en a presque toujours un), avec un tableau de trace. | le tableau de trace rempli |
| **5. Rédiger** | J'écris la réponse propre : du code complet, ou une phrase qui reprend les mots de la question. | la réponse |

⚠️ Dans un sujet, **les questions s'enchaînent** : une fonction écrite à la question 3 sert souvent à la question 5. Si la question dit « en utilisant… », on **appelle** la fonction précédente, on ne la réécrit pas.

### Fiche P5 — Écrire une méthode qui lit ou modifie les attributs

🔎 **Je la reconnais :** « Écrire la méthode `crediter` », « Écrire une méthode `raz` permettant de mettre toutes les coordonnées à zéro », « La méthode `get_annee` renvoie la valeur de l'attribut `annee` », « Écrire la méthode `est_positif` ».

🛠️ **Les 4 sortes de méthodes simples** (repérer laquelle on demande) :

| Sorte | Elle… | Elle se termine par… | Exemple |
|:---|:---|:---|:---|
| **get** (lire) | renvoie un attribut | `return self.attribut` | `get_note` |
| **set** (modifier) | change un attribut avec le paramètre | rien (pas de `return`) | `set_mois`, `changer_role` |
| **test** | répond vrai ou faux | `return` une comparaison | `est_positif`, `est_dispo` |
| **calcul / action** | calcule à partir des attributs, ou les modifie | `return` le résultat, ou rien | `estim_prix`, `crediter`, `avancex` |

**Le squelette :**
```python
    def nom_methode(self, parametres):
        ... self.attribut ...          # on travaille TOUJOURS avec self.
        return ...                     # seulement si la méthode doit renvoyer quelque chose
```

````python
class Compte:
    def __init__(self, titulaire):
        self.titulaire = titulaire
        self.solde = 0

    def get_solde(self):               # get
        return self.solde

    def crediter(self, montant):       # action : modifie, ne renvoie rien
        self.solde = self.solde + montant

    def est_positif(self):             # test
        return self.solde >= 0


c = Compte("Martin")
c.crediter(200)
print(c.get_solde(), c.est_positif())
````

⚠️ **Pièges relevés dans la base :**
- oublier `self.` : `solde = solde + montant` crée une variable locale et ne modifie pas l'objet ;
- oublier `self` comme **premier paramètre** dans le `def` ;
- mettre un `return` dans une méthode qui doit **modifier** (ou l'inverse : modifier un attribut alors qu'on demande de **renvoyer** une valeur) ;
- une méthode « avec limite » (20 points maximum, solde suffisant) : on **teste avant** de modifier.

### Fiche P3 — Créer un objet, appeler ses méthodes

🔎 **Je la reconnais :** « Écrire l'instruction permettant de créer l'objet… », « Parmi les instructions suivantes, laquelle permet de déclarer… », « Écrire un programme permettant : a. la création d'une instance… c. à lancelot d'attaquer le sorcier… », « Écrire la ligne de code permettant de déposer 200 € ».

🛠️ **Les trois écritures à connaître :**

| Je veux… | J'écris | Exemple |
|:---|:---|:---|
| créer un objet | `variable = NomClasse(arguments)` **sans** `self` | `c7 = Carte(7, "coeur")` |
| appeler une méthode | `objet.methode(arguments)` **sans** `self` | `cpt.crediter(200)` |
| lire un attribut | `objet.attribut` (sans parenthèses) | `c1.temps_tot` |

Les arguments sont dans **l'ordre des paramètres du constructeur** (on saute `self`).

⚠️ **Pièges des QCM de la base :** `Carte(self, 7, "coeur")` (le `self` ne s'écrit jamais à l'appel) ; `new Joueur[...]` (c'est d'un autre langage) ; `Joueur{...}` ou une liste `[...]` (ce n'est pas un objet).

### Fiche P6 — Une méthode qui utilise une autre méthode, ou un autre objet

🔎 **Je la reconnais :** « On utilisera les autres méthodes de la classe sans en modifier directement les attributs », « en veillant à ne pas accéder directement aux attributs d'une maison depuis la classe `Antenne` », « prend en paramètre une carte `autre` (de type `Carte`) », « en utilisant les méthodes des classes `Joueur` et `Base` ».

🛠️ **La méthode :**
- appeler une méthode **du même objet** : `self.autre_methode(...)` ;
- utiliser un **autre objet** reçu en paramètre : `autre.attribut` ou, mieux, `autre.methode(...)` ;
- si l'énoncé dit « sans accéder directement aux attributs », on passe **par les méthodes** (`get_...`).

````python
def transferer(self, autre_compte, montant):
    self.solde = self.solde - montant           # interdit si l'énoncé dit « sans modifier les attributs »
    autre_compte.crediter(montant)


def transferer_bien(self, autre_compte, montant):
    self.crediter(-montant)                      # on passe par les méthodes
    autre_compte.crediter(montant)


Compte.transferer = transferer_bien
a, b = Compte("A"), Compte("B")
a.crediter(100)
a.transferer(b, 30)
print(a.get_solde(), b.get_solde())
````

⚠️ **Piège :** dans une méthode qui reçoit `autre`, bien distinguer `self.valeur` (moi) et `autre.valeur` (l'autre). Faire un petit dessin avec deux objets aide beaucoup.

### Fiche P7 — Une fonction qui parcourt une liste d'objets

🔎 **Je la reconnais :** « Écrire une fonction qui prend en paramètre une liste d'objets de la classe `Colis` et qui renvoie… le nombre de… / le poids total… / une nouvelle liste contenant… / le concurrent le plus rapide… / le titre si l'identifiant existe, `None` sinon ».

🛠️ **On reconnaît l'un des 6 modèles**, et on écrit `objet.attribut` dans la boucle :

| Modèle | La question dit… | Squelette |
|:---|:---|:---|
| **compter** | « le nombre de… » | `n = 0` / `for o in liste: if condition: n = n + 1` |
| **additionner** | « le total », « le cumul » | `total = 0` / `total = total + o.attribut` |
| **filtrer** | « une nouvelle liste contenant les… » | `res = []` / `if condition: res.append(o)` |
| **chercher** | « renvoie … si … existe, `None` sinon » | `for o in liste: if condition: return ...` / après la boucle : `return None` |
| **le meilleur** | « le plus rapide », « le plus endetté » | `meilleur = liste[0]` / `if o.attr < meilleur.attr: meilleur = o` |
| **construire** | « renvoie une liste de 52 objets », « crée la liste des maisons » | `res = []` / `res.append(Classe(...))` |

````python
class Colis:
    def __init__(self, ident, poids, etat):
        self.ident = ident
        self.poids = poids
        self.etat = etat


stock = [Colis(1, 12, 'transit'), Colis(2, 30, 'livré'), Colis(3, 5, 'transit')]

total = 0                                   # additionner
for c in stock:
    total = total + c.poids

en_transit = []                             # filtrer
for c in stock:
    if c.etat == 'transit':
        en_transit.append(c)

plus_lourd = stock[0]                       # le meilleur
for c in stock:
    if c.poids > plus_lourd.poids:
        plus_lourd = c

print(total, len(en_transit), plus_lourd.ident)
````

⚠️ **Pièges :** dans le modèle « chercher », le `return None` va **après** la boucle (pas dans un `else`) ; dans « le meilleur », on garde l'**objet** (`meilleur = c`) pour pouvoir renvoyer ce qu'on demande (l'objet, son nom…).

📚 **Origine :** épreuve écrite, **Centres étrangers 2022, jour 2**, exercice 4

*Cet exercice porte sur les structures de données (programmation objet).*

Simon souhaite créer en Python le jeu de cartes « la bataille » pour deux joueurs. À chaque tour, chaque joueur dévoile la carte du haut de son tas ; celui qui présente la carte de plus haute valeur emporte les deux cartes et les place **sous** son tas. Les valeurs des cartes sont, de la plus forte à la plus faible : As, Roi, Dame, Valet, 10, 9, 8, 7, 6, 5, 4, 3 et 2.

### Question 1.a

🏷️ *Type de question :* **P2** Compléter le constructeur `__init__`

Simon a écrit la classe Python `Carte` suivante, ayant deux attributs `valeur` et `couleur`, et dont le constructeur prend deux arguments : `val` et `coul`. Recopier et compléter les `…` des lignes 3 et 4.

Dans cet exercice, on étudie une méthode de chiffrement de chaînes de caractères alphabétiques. Pour des raisons historiques, cette méthode de chiffrement est appelée « code de César ». On considère que les messages ne contiennent que les lettres capitales de l'alphabet `"ABCDEFGHIJKLMNOPQRSTUVWXYZ"` et la méthode de chiffrement utilise un nombre entier fixé appelé la clé de chiffrement.

*La classe `CodeCesar` donnée par le sujet (à exécuter) :*

````python
class CodeCesar:

    def __init__(self, cle):
        self.cle = cle
        self.alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    def decale(self, lettre):
        num1 = self.alphabet.find(lettre)
        num2 = num1 + self.cle
        if num2 >= 26:
            num2 = num2 - 26
        if num2 < 0:
            num2 = num2 + 26
        nouvelle_lettre = self.alphabet[num2]
        return nouvelle_lettre
````
